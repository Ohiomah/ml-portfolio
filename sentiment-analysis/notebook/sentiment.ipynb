```python
import pandas as pd
import re
import nltk

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
```

```python
df = pd.read_csv("IMDB Dataset.csv")
print(df.head())
df['sentiment'].value_counts()
```

```python
print(df['review'][0])
print(df['sentiment'][0])
```

```python
def clean_text(review) :
    review = re.sub("<br />", " ", review)
    review = review.lower()
    review = re.sub("\W", " ", review)
    review = re.sub("  ", " ", review)

    return review
```

```python
df['clean_review'] = df['review'].apply(clean_text)
print(df.head())
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
print(stop_words)
```

```python
def remove_stopwords(text):
    words = text.split()
    filtered = [w for w in words if w not in stop_words]
    return " ".join(filtered)

df['clean_review'] = df['clean_review'].apply(remove_stopwords)
print(df.head())
```

```python
tfidf = TfidfVectorizer(max_features=5000)
result = tfidf.fit_transform(df['clean_review'])
print(result.shape)

df['sentiment'] = df['sentiment'].map({'positive' : 1, 'negative' : 0})
print(df.head())
```

```python
y = df['sentiment']
x = result

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
model = LogisticRegression(max_iter=10000, random_state=0)
model.fit(x_train, y_train)

prediction = model.predict(x_test)
accuracy_score(y_test, prediction)
```

```python
model2 = MultinomialNB()
model2.fit(x_train, y_train)

prediction2 = model2.predict(x_test)
accuracy_score(y_test, prediction2)
```

```python
model3 = LinearSVC()
model3.fit(x_train, y_train)

prediction3 = model3.predict(x_test)
accuracy_score(y_test, prediction3)
```

```python
my_review = ["This movie was absolutely terrible, I hated every minute of it"]
my_review_clean = [remove_stopwords(clean_text(my_review[0]))]
my_review_tfidf = tfidf.transform(my_review_clean)
prediction4 = model.predict(my_review_tfidf)
print("positive" if prediction4[0] == 1 else "negative")

my_review = ["This movie was absolutely brilliant, I loved every minute of it"]
my_review_clean = [remove_stopwords(clean_text(my_review[0]))]
my_review_tfidf = tfidf.transform(my_review_clean)
prediction4 = model.predict(my_review_tfidf)
print("positive" if prediction4[0] == 1 else "negative")

my_review = ["This movie was so bad it was actually good, I couldn't stop watching"]
my_review_clean = [remove_stopwords(clean_text(my_review[0]))]
my_review_tfidf = tfidf.transform(my_review_clean)
prediction4 = model.predict(my_review_tfidf)
print("positive" if prediction4[0] == 1 else "negative")
```

```python
import joblib

joblib.dump(model, 'sentiment_model.joblib')
joblib.dump(tfidf, 'tfidf_vectorizer.joblib')
```

```python
from transformers import pipeline

classifier = pipeline('sentiment-analysis')

result = classifier("This movie was so bad it was actually good, I couldn't stop watching")
print(result)

reviews = [
    "This movie was absolutely terrible, waste of my time",
    "Not bad at all, I actually enjoyed it",
    "The acting was good but the plot was boring",
    "I can't believe how amazing this film was"
]

for review in reviews:
    result = classifier(review)
    print(f"{result[0]['label']} ({result[0]['score']:.2%}) — {review}")
```
